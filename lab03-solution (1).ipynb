### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
for n in [100, 173, 530, 746, 911, 1024]:
    R = Integers(n)
    ds = divisors(n)
    orders = [x.additive_order() for x in R]
    print("\nn =", n)
    print("1. Дільники:", ds)
    print("2–3. Порядок d | кількість | phi(d) | збіг")
    for d in ds:
        count = orders.count(d)
        print(d, count, euler_phi(d), count == euler_phi(d))
    print("4. Усі підгрупи (вказано твірний та елементи):")
    subgroups = []
    for d in ds:
        generator = R(n // d)
        H = [ZZ(j * generator) for j in range(d)]
        subgroups.append(H)
        print("Порядок", d, "; твірний", generator, "; H =", H)
    print("5. Кількість підгруп:", len(subgroups))
    print("6. tau(n) =", number_of_divisors(n))
    print("Збіг:", len(subgroups) == number_of_divisors(n))
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
for n in [5, 6, 10, 12, 16, 17, 20]:
    D = DihedralGroup(n)
    orders = [g.order() for g in D]
    print("\nn =", n, "; |D_n| =", D.order())
    expected = n if n % 2 else n + 1
    print("1. Елементів порядку 2:", orders.count(2))
    print("Теоретично:", expected, "; збіг:", orders.count(2) == expected)
    print("2. Порядок d | кількість | теоретично | збіг")
    for d in divisors(2 * n):
        if d > 2:
            count = orders.count(d)
            expected = euler_phi(d) if n % d == 0 else 0
            print(d, count, expected, count == expected)
    missing = [d for d in divisors(2 * n) if d < 2 * n and d not in orders]
    print("Дільники d < 2n, для яких немає елементів:", missing)
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
n = ZZ(1102158078129785185997827239806)
R = Integers(n)
print("1. Факторизація n:", factor(n))
print("2. Група циклічна:", R.multiplicative_group_is_cyclic())
group_order = euler_phi(n)
print("Порядок групи phi(n) =", group_order)
print("Факторизація порядку:", factor(group_order))

a = ZZ(1)
while gcd(a, n) != 1 or R(a).multiplicative_order() != group_order:
    a += 1
base = R(a)
print("3. Найменший натуральний твірний a =", a)
print("4. Порядок твірного:", base.multiplicative_order())
print("Порядок дорівнює phi(n):", base.multiplicative_order() == group_order)

set_random_seed(2026)
elements = []
while len(elements) < 3:
    g = R.random_element()
    if g.is_unit() and g not in elements:
        elements.append(g)
print("5. Випадкові елементи:", elements)
print("6. Дискретні логарифми та перевірки:")
for i, g in enumerate(elements, 1):
    k = g.log(base)
    print("g_" + str(i), "=", g)
    print("k =", k, "; a^k =", base ** k, "; збіг:", base ** k == g)
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:
1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?

```sage
A = matrix(ZZ, [[1, 0, 1], [0, 1, 0], [0, 0, 1]])
B = matrix(ZZ, [[1, 0, 0], [0, 1, 1], [0, 0, 1]])
I = identity_matrix(ZZ, 3)
print("A*B = B*A:", A * B == B * A)

m, r, p, q, k = var("m r p q k")
X = matrix(SR, [[1, 0, m], [0, 1, r], [0, 0, 1]])
Y = matrix(SR, [[1, 0, p], [0, 1, q], [0, 0, 1]])
N = X - I
print("Загальний вигляд X(m,r):")
show(X)
print("Перевірка N^2 = 0:", N ** 2 == zero_matrix(SR, 3))
print("Добуток X(m,r)*X(p,q):")
show(X * Y)
print("X(m,r)^k = I + k*N (формула випливає з N^2=0):")
show(I + k * N)
print("Приклади степенів A^j*B^t для цілих j,t:")
for j, t in [(-2, 3), (0, 0), (1, -1), (3, 2)]:
    expected = matrix(ZZ, [[1, 0, j], [0, 1, t], [0, 0, 1]])
    print("j =", j, "; t =", t, "; збіг:", A ** j * B ** t == expected)
    show(A ** j * B ** t)
print("1. Група нескінченна: усі A^j різні.")
print("2. Група абелева: твірні A і B комутують.")
print("3. Порядок I дорівнює 1; усіх інших елементів — нескінченність.")
print("4. Усі елементи — X(m,r), де m,r є цілими числами.")
print("5. Група не циклічна: одна пара (m,r) не породить усі пари цілих чисел.")
print("6. G ізоморфна Z x Z відносно додавання.")
```
