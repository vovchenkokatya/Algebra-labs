### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
print("Адитивна група Z10")
G1 = AdditiveAbelianGroup([10])
a1, b1 = G1([2]), G1([3])
print("Елементи a, b:", a1, ",", b1)
print("Добуток a*b:", a1 + b1)
print("Обернені a^-1, b^-1:", -a1, ",", -b1)
print("Перевірка a + (-a) == e:", a1 + (-a1) == G1.zero())
print("Порядок підгрупи, породженої a та b:", G1.submodule([a1, b1]).order())
print("Елементи, обернені самі до себе (a == -a):", [x for x in G1 if x == -x])
is_abelian_G1 = all(x + y == y + x for x in G1 for y in G1)
print("Перевірка абелевості:", is_abelian_G1)
if not is_abelian_G1:
    print("Не є абеелевою! a+b != b+a:", a1 + b1 != b1 + a1)
```

```sage
print("Мультиплікативна група Z10^*")
G2 = Zmod(10).unit_group()
e2 = G2.list()
a2, b2 = e2[1], e2[2] 
print("Елементи a, b (в Z_10):", a2.value(), ",", b2.value())
print("Добуток a*b:", (a2 * b2).value())
print("Обернені a^-1, b^-1:", (a2^-1).value(), ",", (b2^-1).value())
print("Перевірка a * a^-1 == e:", a2 * a2^-1 == G2.one())
print("Порядок підгрупи, породженої a та b:", G2.subgroup([a2, b2]).order())
print("Елементи, обернені самі до себе:", [x.value() for x in G2 if x^-1 == x])
is_abelian_G2 = all(x * y == y * x for x in G2 for y in G2)
print("Перевірка абелевості:", is_abelian_G2)
if not is_abelian_G2:
    print("Не є абелевою! a*b != b*a:", a2 * b2 != b2 * a2)
```

```sage
print("A4")
G3 = AlternatingGroup(4)
a3, b3 = G3("(1,2,3)"), G3("(1,2)(3,4)")
print("Елементи a, b:", a3, ",", b3)
print("Добуток a*b:", a3 * b3)
print("Обернені a^-1, b^-1:", a3^-1, ",", b3^-1)
print("Перевірка a * a^-1 == e:", a3 * a3^-1 == G3.one())
print("Порядок підгрупи, породженої a та b:", G3.subgroup([a3, b3]).order())
print("Елементи, обернені самі до себе:", [x for x in G3 if x^-1 == x])
is_abelian_G3 = all(x * y == y * x for x in G3 for y in G3)
print("Перевірка на абелевість:", is_abelian_G3)
if not is_abelian_G3:
    print("Не є абелевою! a*b != b*a:", a3 * b3 != b3 * a3)
```

```sage
print("GL2(Z3)")
G4 = GL(2, GF(3))
a4, b4 = G4([[1,1],[0,1]]), G4([[1,0],[1,1]])
print("Елементи a, b:\n", a4, "\n\n", b4)
print("Добуток a*b:\n", a4 * b4)
print("Обернені a^-1, b^-1:\n", a4^-1, "\n\n", b4^-1)
print("Перевірка a * a^-1 == I:", a4 * a4^-1 == G4.one())
print("Порядок підгрупи, породженої a та b:", G4.subgroup([a4, b4]).order())
print("Елементи, обернені самі до себе:")
print([x for x in G4 if x^-1 == x])
is_abelian_G4 = all(x * y == y * x for x in G4 for y in G4)
print("Перевірка на абелевість:", is_abelian_G4)
if not is_abelian_G4:
    print("Не є абелевою! a*b != b*a:", a4 * b4 != b4 * a4)
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
print("Адитивна група Z24")
G1 = CyclicPermutationGroup(24)
subgroups_G1 = G1.subgroups()
print("Список усіх підгруп:")
for H in subgroups_G1:
    print("  ", H)

iso_classes_G1 = []
for H in subgroups_G1:
    if not any(H.is_isomorphic(K) for K in iso_classes_G1):
        iso_classes_G1.append(H)
print("Кількість підгруп з точністю до ізоморфізму:", len(iso_classes_G1))

orders_G1 = [H.order() for H in subgroups_G1]
print("Порядки всіх підгруп:", orders_G1)
print("Чи ділить порядок кожної підгрупи порядок групи?", all(G1.order() % o == 0 for o in orders_G1))
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
print("=== ГРАФ КЕЛІ ДЛЯ D4 ===")
D4 = DihedralGroup(4)
print(f"Порядок групи: {D4.order()}")


S = D4.gens()
print(f"Система твірних: {S}")


cg = D4.cayley_graph(generators=S)
print(f"Граф Келі має {cg.num_verts()} вершин та {cg.num_edges()} ребер")


plot = cg.plot(color_by_label=True, vertex_size=500, vertex_labels=True)
plot.show()
print("\n" + "="*50 + "\n")
```
