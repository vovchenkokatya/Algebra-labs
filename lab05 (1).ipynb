### 1. Структура абелевих груп

Розглянемо абелеві групи порядку 400.

ℤ₄₀₀, ℤ₁₆ × ℤ₂₅, ℤ₂₀ × ℤ₂₀, ℤ₄ × ℤ₁₀₀, ℤ₂ × ℤ₁₀ × ℤ₂₀, ℤ₅₀₅⁎.

Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних p-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```sage
from collections import Counter
from itertools import combinations, product

groups = [
    ("Z400", AbelianGroup([400])),
    ("Z16 × Z25", AbelianGroup([16, 25])),
    ("Z20 × Z20", AbelianGroup([20, 20])),
    ("Z4 × Z100", AbelianGroup([4, 100])),
    ("Z2 × Z10 × Z20", AbelianGroup([2, 10, 20])),
    ("Z505*", Zmod(505).unit_group())
]

def elementary_divisors(G):
    powers = [p**k for m in G.gens_orders() for p, k in factor(m)]
    return tuple(sorted(powers))

types = {}
for name, G in groups:
    parts = elementary_divisors(G)
    types[name] = parts
    orders = Counter(g.order() for g in G)
    assert G.order() == sum(orders.values()) == 400

    print("\nГрупа", name)
    print("Елементарні дільники:", parts)
    print("Розклад:", " × ".join("Z" + str(d) for d in parts))
    print("Порядок елемента → кількість:", dict(sorted(orders.items())))

print("\nІзоморфні пари:")
for a, b in combinations(types, 2):
    if types[a] == types[b]:
        print(a, "≅", b)
```

### 2. Одна абелева група порядку n
1. Реалізуйте функцію, яка для заданого натурального числа n обчислює кількість попарно неізоморфних абелевих груп порядку n.
2. Знайдіть перші сто таких натуральних чисел n, для яких існує рівно одна абелева група порядку n. Для кожного знайденого n проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких n ∈ ℕ існує рівно одна (з точністю до ізоморфізму) абелева група порядку n.


```sage
def abelian_group_count(n):
    return prod(number_of_partitions(k) for p, k in factor(n))

numbers = []
n = 1
while len(numbers) < 100:
    if abelian_group_count(n) == 1:
        numbers.append(n)
    n += 1

print("n | розклад на прості множники")
for n in numbers:
    print(n, "|", factor(n))

all_squarefree = all(k == 1 for n in numbers for p, k in factor(n))
print("\nКількість знайдених чисел:", len(numbers))
print("Усі степені простих множників дорівнюють 1:", all_squarefree)
```

Нехай p₁, p₂, … — прості множники числа n, а k₁, k₂, … — їхні показники. Для кожного простого множника pᵢ можливі абелеві групи порядку pᵢ у степені kᵢ відповідають розбиттям числа kᵢ. Тому загальна кількість груп дорівнює P(k₁) · P(k₂) · …, де P(kᵢ) — число розбиттів kᵢ.

Гіпотеза за результатами обчислень — існує рівно одна абелева група порядку n, якщо й лише якщо n вільне від квадратів. Це випливає з формули, бо P(1)=1, а при k ≥ 2 є принаймні два розбиття, k і 1+…+1. Для n=1 порожній добуток дорівнює 1, і група тривіальна.

### 3. Дільники нуля та одиниці в скінченних кільцях
Розгляньте скінченні кільця ℤ₃₀, ℤ₁₂ × ℤ₁₅, M₃(ℤ₃) та для кожного з них виконайте наступні кроки:
1. Знайдіть дільники нуля та їх кількість.
2. Знайдіть дільники одиниці та їх кількість.
3. Порівняйте отримані результати та зробіть висновок про взаємозв'язок між дільниками нуля та дільниками одиниці в скінченних кільцях.


```sage
R = Zmod(30)
units_30 = [a for a in R if a.is_unit()]
zero_divisors_30 = [a for a in R if a != 0 and not a.is_unit()]
print("Z30")
print("Дільники нуля:", zero_divisors_30)
print("Дільники одиниці:", units_30)
print("Кількості:", len(zero_divisors_30), len(units_30))

R12, R15 = Zmod(12), Zmod(15)
units_product = []
zero_divisors_product = []
for a, b in product(R12, R15):
    if a.is_unit() and b.is_unit():
        units_product.append((a, b))
    elif (a, b) != (0, 0):
        zero_divisors_product.append((a, b))

print("\nZ12 × Z15")
print("Дільники нуля:", zero_divisors_product)
print("Дільники одиниці:", units_product)
print("Кількості:", len(zero_divisors_product), len(units_product))

F = GF(3)
M = MatrixSpace(F, 3)
units_matrices = []
zero_divisors_matrices = []
for entries in product(F, repeat=9):
    A = M(list(entries))
    if A.det() != 0:
        units_matrices.append(A)
    elif A != M.zero():
        zero_divisors_matrices.append(A)

print("\nM3(Z3)")
print("Дільники нуля: усі ненульові A з det(A) = 0")
print("Дільники одиниці: усі A з det(A) ≠ 0")
print("Кількості:", len(zero_divisors_matrices), len(units_matrices))
assert len(units_matrices) == (27 - 1) * (27 - 3) * (27 - 9)
```

### 4. Факторкільця кільця многочленів

Нехай f(x)=x⁴+x+2. Виконайте наступні кроки для кожного з кілець R=ℚ, ℤ₄, ℤ₅:

1. Побудуйте ідеал I=(f(x)) у кільці R[x].
2. Побудуйте факторкільце R[x]/I та знайдіть його порядок.
3. Визначте, чи є факторкільце R[x]/I полем.
4. Знайдіть дільники нуля у факторкільці R[x]/I.
5. Виділіть усі лінійні множники f(x) над R та порівняйте з дільниками нуля у факторкільці R[x]/I.
6. Перевірте, чи елемент x+I ∈ R[x]/I є оборотним.



```sage
P2 = PolynomialRing(GF(2), "x")
x2 = P2.gen()
f2 = x2**4 + x2
zero_divisors_quotients = {}

for R in [QQ, Zmod(4), GF(5)]:
    P = PolynomialRing(R, "x")
    x = P.gen()
    f = x**4 + x + 2
    I = P.ideal(f)
    Q = P.quotient(I, "a")
    a = Q.gen()

    print("\nR =", R)
    print("Ідеал:", I)
    print("Факторкільце:", Q)
    print("Порядок:", Q.cardinality())

    if R == Zmod(4):
        factors = (x - 1, x - 2, x**2 + 3*x + 3)
        assert prod(factors) == f
        print("Розклад f:", factors)
        print("Корені:", [r for r in R if f(r) == 0])
    else:
        print("Розклад f:", f.factor())
        print("Корені:", f.roots(multiplicities=False))

    is_field = R.is_field() and f.is_irreducible()
    print("Поле:", is_field)

    zero_divisors = []
    if R.is_finite():
        for coefficients in product(R, repeat=4):
            g = P(list(coefficients))
            if g == 0:
                continue
            if R == Zmod(4):
                reduced = P2([int(c) for c in g.list()])
                is_unit = reduced.gcd(f2) == 1
            else:
                is_unit = g.gcd(f) == 1
            if not is_unit:
                zero_divisors.append(Q(g))
    zero_divisors_quotients[R] = zero_divisors
    print("Кількість ненульових дільників нуля:", len(zero_divisors))
    if zero_divisors:
        print("Перші 10:", zero_divisors[:10])

    if R(2).is_unit():
        inverse = -Q(R(2)**(-1)) * (a**3 + 1)
        print("a оборотний, a^(-1) =", inverse)
        print("Перевірка a*a^(-1) = 1:", a * inverse == Q.one())
    else:
        b = 2 * (a**3 + 1)
        print("a не є оборотним. Ненульовий елемент b =", b)
        print("Перевірка a*b = 0:", a * b == Q.zero())
```
