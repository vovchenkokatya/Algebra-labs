### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
G = SymmetricGroup(4)
H1 = G.subgroup([G("(1,2,3,4)")])
H2 = G.subgroup([G("(1,3)"), G("(1,2,3,4)")])
H3 = G.subgroup([G("(1,2)(3,4)"), G("(1,3)(2,4)")])

for name, H in [("H1", H1), ("H2", H2), ("H3 = K4", H3)]:
    print("\nПідгрупа", name)
    print("Елементи:", list(H))
    print("Порядок:", H.order())
    print("Індекс:", G.order() // H.order())

    left = G.cosets(H, side="left")
    right = G.cosets(H, side="right")
    print("Ліві класи aH:")
    for C in left:
        print(C)
    print("Праві класи Ha:")
    for C in right:
        print(C)

    normal = H.is_normal(G)
    print("Нормальна:", normal)
    if normal:
        Q = G.quotient(H)
        print("Порядок факторгрупи:", Q.order())
        print("Структура факторгрупи:", Q.structure_description())
        print("Ізоморфна S3:", Q.is_isomorphic(SymmetricGroup(3)))
        print("Таблиця Келі факторгрупи:")
        print(Q.cayley_table(names="elements"))
    else:
        for a in G:
            aH = {a * h for h in H}
            Ha = {h * a for h in H}
            if aH != Ha:
                print("Приклад a =", a)
                print("aH =", aH)
                print("Ha =", Ha)
                break
```

### 2. Нормальні підгрупи в діедральних групах

Розгляньте діедральні групи $D_n$ для значень $n\in\{3,4,\dots,30\}$ і виконайте такі кроки:
1. Побудуйте групу $D_n$.
2. Визначте кількість нормальних підгруп у $D_n$.
3. Складіть перелік порядків усіх нормальних підгруп.
4. Визначте алгебраїчну структуру кожної нормальної підгрупи.
5. Знайдіть закономірність і виведіть гіпотезу про кількість та структуру нормальних підгруп в $D_n$.
6. Сформулювати гіпотезу: Скільки нормальних підгруп в $D_n$ та яка їхня структура? Спрогнозувати кількість нормальних підгруп у $D_{470448}$ без виконання обчислень. 

```sage
for n in range(3, 31):
    D = DihedralGroup(n)
    normals = sorted(D.normal_subgroups(), key=lambda H: H.order())

    print("\nD_" + str(n))
    print("Порядок групи:", D.order())
    print("Кількість нормальних підгруп:", len(normals))
    print("Їхні порядки:", [H.order() for H in normals])

    for i, H in enumerate(normals, 1):
        print(str(i) + ". Порядок: " + str(H.order())
              + "; структура: " + H.structure_description())

    expected = number_of_divisors(n) + (3 if n % 2 == 0 else 1)
    print("Кількість за формулою:", expected)
    print("Збігається:", len(normals) == expected)

print("\nВисновки:")
print("Непарне n: tau(n) + 1 нормальних підгруп.")
print("Парне n: tau(n) + 3 нормальних підгруп.")
print("Усі підгрупи поворотів нормальні та циклічні.")
print("За парного n є ще дві підгрупи індексу 2.")
print("Сама група також нормальна в собі.")
print("470448 = 2^4 * 3^5 * 11^2.")
print("Для D_470448: (4+1)*(5+1)*(2+1)+3 =", (4+1)*(5+1)*(2+1)+3)
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
from sage.libs.gap.libgap import libgap

F = GF(3)
G = GL(2, F)
S = SymmetricGroup(4)
vectors = [vector(F, [1, 0]), vector(F, [0, 1]),
           vector(F, [1, 1]), vector(F, [1, 2])]

line_index = {}
for i, v in enumerate(vectors, 1):
    line_index[tuple(v)] = i
    line_index[tuple(2 * v)] = i

def phi(A):
    images = []
    for v in vectors:
        w = v * A.matrix()
        images.append(line_index[tuple(w)])
    return S(images)

elements = list(G)
print("Порядок GL(2,3):", G.order())
print("Дія твірних на чотирьох прямих:")
for A in G.gens():
    print(A.matrix())
    print("phi(A) =", phi(A))

is_hom = all(phi(A * B) == phi(A) * phi(B) for A in elements for B in elements)
print("Перевірка phi(A*B) = phi(A)*phi(B) для всіх пар:", is_hom)

kernel_elements = [A for A in elements if phi(A) == S.one()]
N = G.subgroup(kernel_elements)
image = S.subgroup([phi(A) for A in elements])
print("Порядок ядра:", N.order())
print("Матриці ядра:")
for A in N:
    print(A.matrix())
print("Порядок образу:", image.order())
print("Образ дорівнює S4:", image == S)

Q = libgap.FactorGroup(libgap(G), libgap(N))
print("Порядок факторгрупи:", Q.Size())
print("Структура факторгрупи:", Q.StructureDescription())
iso = libgap.IsomorphismGroups(Q, libgap(image))
print("Факторгрупа ізоморфна образу:", iso != libgap.fail)
```
