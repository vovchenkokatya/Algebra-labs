### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
var('n')  
n = -2

if n > 0:
    print("n - додатне число")
elif n < 0:
    print("n - від’ємне число")
else:
    print("n = 0")
```

```sage
for i in range(1, 11):
    print(i)
```

```sage
def prime(n):
    if n <= 1:
        return False
    for i in range(2, int(sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

print(prime(17))
```
